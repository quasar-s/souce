### 딕셔너리 자료형
- {key1:value1, key2:value2}
- key값 중복 불가, value 값 중복 가능

````python
dict1 = {"name":"kim", "age":22}
dict2 = {1:"kim", 2:22, 3:"서울"}
dict3 = {"arr":[1,2,3,4,5]}
````

````python
# Kim 출력
print(dict1["name"])
# print(dict1["addr"]): Error
print(dict1.get("name")) #추천
print(dict1.get("addr"))#: None
# 서울 출력
print(dict2[3])
print(dict2.get(3))
````

````python
# 삭제 del

del dict1["age"]
print(dict1)
````

````python
# 추가 딕셔너리[key] = value
# birth : '1999-09-12'
dict1['birth'] = '1999-09-12'
print(dict1)
````

````python
# rank = [15,17,19]
dict1['rank'] = [15,17,19]
print(dict1)
````

````python
# dict.keys(): 키값만 가져오기
# dict.values(): 값만 가져오기
# dict.items(): 키,값 쌍으로 가져오기
print(dict1.keys())
print(dict1.values())
print(dict1.items())
````

````python
# in: b 안에 a가 있는가?
print('사과' in ['사과','포도','바나나'])

# key가 포함 되어 있는가?
print('사과' in dict1)
print('name' in dict1)

# not in
print('사과' not in ['사과','포도','바나나'])
````

````python
# input()을 통해 얻는 데이터 = 문자열
#num = int(input("당신의 점수를 입력하세요: "))
num = input("당신의 점수를 입력하세요: ")
grade = "D"

if not num.strip():
    num = 0
    print("입력이 정상적이지 않습니다.")
else:
    num = int(num)
    if num >= 90:
        grade = "A"
        print("A")
    elif num >= 80:
        grade = "B"
        print("B")
    elif num >= 70:
        grade = "C"
        print("C")
    else:
        grade = "D"
        print("D")

# 점수가 90이고 평가 등급이 A라면 합격입니다. 그외에는 불합격입니다.

if num >= 90 and grade == "A":
    print("합격입니다!")
else:
    print("불합격입니다.")
````
