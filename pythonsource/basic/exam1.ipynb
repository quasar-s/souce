````python
# 주차장 프로그램
# 5대의 주차 공간
# A, B, C, D, E

# 메뉴 제공
# [1] 주차 | [2] 출차 | [3] 종료

# 1번 선택 => A 자동차 들어감. 주차장 상태 ==> ['A']
# 1번 선택 => B 자동차 들어감. 주차장 상태 ==> ['A', 'B']
# 2번 선택 => B 자동차 나감. 주차장 상태 ==> ['A']

#5대 까지만 가능하기 때문에 6번째 차량이 들어오려고 하면 '주차장 만차'

top, car_name = 0, "A"

parking_lot = []

while True:
    menu = int(input("[1] 주차 | [2] 출차 | [3] 종료"))
    
    if menu == 1:
        if 5 > top:
            parking_lot.append(car_name)
            print(f"{car_name}자동차 들어옴. 현재 주차장 상태:{parking_lot}")

            top+=1
            car_name = chr(ord(car_name)+1)
        else:
            print("현재 주차장이 만차입니다.")
    elif menu == 2:
        if top > 0:
            out_car = parking_lot.pop()
            print(f"{out_car}자동차 나감. 현재 주차장 상태:{parking_lot}")

            top-=1
            car_name = chr(ord(car_name)-1)
        else:
            print("현재 주차장이 공석입니다.")
    elif menu == 3:
        print("프로그램 종료")
        break
    else:
        print("다시 입력하시기바랍니다.")
````

### 클래스
- 함수, 변수를 포함한 형태
- 객체지향

````python
# 클래스 생성
class Calculator:
    def __init__(self):
        self.result =0
    
    # 메서드
    def add(self, num):
        self.result += num
        return self.result

    def sub(self, num):
        self.result -= num
        return self.result
````

````python
class FourCal:
    # 생성자
    def __init__(self, first, second):
        self.first = first
        self.second = second
    
    # 메서드
    def add(self):
        return self.first+self.second
    def sub(self):
        return self.first-self.second
    def mul(self):
        return self.first*self.second
    def div(self):
        return self.first//self.second
    
    def __str__(self):
        return f"{self.first}, {self.second}"
````

````python
# __init__() 자동호출
calc1 = FourCal(5,3)

# 사칙연산
print(calc1.add())
print(calc1.sub())
print(calc1.mul())
print(calc1.div())
print(calc1)
````

````python
# 상속

class MoreFourCal(FourCal):
    def pow(self):
        return self.first ** self.second
````

````python
more = MoreFourCal(4,2)

print(more.add())
print(more.sub())
print(more.mul())
print(more.div())
print(more.pow())
print(more)
````

````python
# 오버라이딩: 부모 클래스 메소드 재정의

four1 = FourCal(4, 0)
# four1.div()
````

````python
# super(): 부모
# self : 자기자신
class SafeFourCal(FourCal):
    # 오버라이딩: 자식에서 재정의
    def div(self):
        if self.second == 0:
            return 0
        else:
            return self.first/self.second 
````

````python
four1 = SafeFourCal(4, 0)
four1.div()
````
