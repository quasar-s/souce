```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model="WhitePeak/bert-base-cased-Korean-sentiment", top_k=None) # 감정 분석

classifier("이 치킨은 맛도 없고 이상하다")
```

```python
classifier = pipeline("sentiment-analysis", model="snunlp/KR-FinBert-SC", top_k=None) # 감정 분석
texts = """\
이 치킨은 맛도 없고 이상하다.
나는 행복해.
배고파."""
texts = texts.splitlines()
for item in texts:
    print(classifier(item))
```

```python
classifier = pipeline("sentiment-analysis", model="nlptown/bert-base-multilingual-uncased-sentiment") # 감정 분석
classifier("이 치킨은 맛도 없고 이상하다")
```
