### LLM 프레임 워크

```python
# 라이브러리 로드
from langchain_ollama import ChatOllama
from langchain_ibm import ChatWatsonx
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser, PydanticOutputParser
from pydantic import BaseModel, Field
from typing import Literal
from dotenv import load_dotenv
import os
```

```python
#.env 내용 가죠오기
load_dotenv()

apikey = os.getenv("WATSONX_API_KEY")
project_id = os.getenv("WATSONX_PROJECT_ID")
watsonx_ai_url = os.getenv("WATSONX_URL")
```

### LangChain+watsonx 사용하기

```python
watson_llm = ChatWatsonx(
    model_id="ibm/granite-4-h-small",
    url = f"{watsonx_ai_url}",
    api_key = f"{apikey}",
    project_id=f"{project_id}",
    params = {
    "max_tokens": 2000
    }
)

response = watson_llm.invoke("생성형 AI 설명해줘")
print(response.content)
```

```python
qwen_llm = ChatOllama(model="qwen3.5:4b")

response = qwen_llm.invoke("생성형 AI 설명해줘")
print(response.content)
```

```python
exaone_llm = ChatOllama(model="exaone3.5:2.4b")

response = exaone_llm.invoke("생성형 AI 설명해줘")
print(response.content)
```

### ChatPromptTemplate
- 채팅 메세지 형식
- from_template() : human 메세지만 생성
- from_messages() : 여러 메세지를 구성할 수 있음

```python
template = ChatPromptTemplate.from_template("""\
다음 질문에 답변하시오.

질문:
{question}
""")

formatted_prompt = template.format(question = "SQL Injection이란?")
response = watson_llm.invoke(formatted_prompt)
print(response.content)
```

```python
template = ChatPromptTemplate.from_messages(
    [
        ("system", "당신은 {role} 전문가입니다. {language}로 답변하세요"),
        ("human", "{question}"),
    ]
)

formatted_prompt = template.format(role = "Python",language="한국어",question = "리스트 컴프리헨션이란")
response = watson_llm.invoke(formatted_prompt)
print(response.content)
```

### LCEL 파이프라인
- LCEL(LangChain Expression Language) : \ 연산자로 컴포턴트를 연결하는 파이프라인 문법
    - ex) prompt | llm | parser => 왼쪽에서 오른 쪽으로 데이터가 흐르게 됨
- invoke() : 단일 동기 호출 / stream() : 스트리밍 / batch() : 여러 입력 병렬 처리

```python
prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "당신은 {role} 전문가입니다."),
        ("human", "{question}"),
    ]
)

chain = prompt | watson_llm

# formatted_prompt = prompt.format(role = "보안",question = "xxs란?")
response = chain.invoke({
    "role" : "보안",
    "question" : "xxs란?"
})
print(response.content)
```
