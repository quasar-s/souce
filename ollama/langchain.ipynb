### LCEL 파이프라인
- LCEL(LangChain Expression Language) : \ 연산자로 컴포턴트를 연결하는 파이프라인 문법
    - ex) prompt | llm | parser => 왼쪽에서 오른쪽으로 데이터가 흐르게 됨
    - RunnableSequence 생성
        - RunnablePassthrough : 입력을 그대로 통과시키며 다른 키와 병합할 때 사용
        - RunnableLambda : 체인을 일반 파이썬 함수와 연결
        - RunnableParallel :여러체인을 동시에 실행하고 결과를 dict 병합할 때 사용
        - .assign() : 체인 중간에 새 키 추가
- invoke() : 단일 동기 호출 / stream() : 스트리밍 / batch() : 여러 입력 병렬 처리

```python
# 감정, 점수, 이유, 키워드 3개
# Pydantic 모델 정의
class SentimentResult(BaseModel):
    sentiment:Literal["positive", "negative", "neutral"]
    score:float = Field(ge=0.0, le=1.0, description="감정 강도")
    reason: str = Field(description="판단 근거 한 문장")
    keywords:list[str] = Field(description="핵심 키워드 3개 이내")

# 파서 정의
pydatic_parser = PydanticOutputParser(pydantic_object=SentimentResult)

# 프롬프트 작성
# {format_instructions} : pydatic_parser의 필드를 참고하여 작성해줘
prompt = ChatPromptTemplate.from_messages([
    ('system', '감정 분석 전문가입니다.\n\n {format_instructions}'),
    ('human', '다음 텍스트의 감정을 분석 하세요 \n {text}'),
]).partial(format_instructions = pydatic_parser.get_format_instructions())

chain = prompt | watson_llm | pydatic_parser
result = chain.invoke({
    "text" : "오늘 최악의 하루였습니다. 모든 게 잘못되었습니다."
})

print(type(result)) 
print(result.sentiment) 
print(result.score) 
print(result.reason) 
print(result.keywords) 
```

```python
texts = [
    "이 제품 정말 별로네요. 돈 낭비 였습니다.",
    "그냥 폄범한 하루였습니다. 특별한 건 없었습니다.",
    "와, 생각보다 훨씬 맛있었어요! 강력 추천드립니다.",
    "서비스가 별로고 맛도 그닥이었습니다.",
]

chain = prompt | watson_llm | pydatic_parser
results = chain.batch([{'text':t}  for t in texts])

print("==== 감정 분석 결과 ====")

for text, r in zip(texts, results): 
    bar = '' * int(r.score * 10) + '' * (10 - int(r.score * 10))
    emoji = {'positive':'🤩','negative':'👿','neutral':'🤔'}[r.sentiment]
    print(f'{emoji}[{r.sentiment:8s}]{bar}{r.score:.2f}')
    print(f'입력 : {text[:30]}....')
    print(f'이유 : {r.reason}')
```

```python
class AdvancedResult(BaseModel):
    sentiment:Literal["positive", "negative", "neutral"]
    score:float = Field(ge=0.0, le=1.0, description="감정 강도")
    category: Literal['product', 'service', 'food', 'general']= Field(description="리뷰 카테고리")
    summary:str =Field(description="한 줄 요약")
    action:str =Field(description="비즈니스 조치 방향")

# 파서 정의
pydatic_parser = PydanticOutputParser(pydantic_object=AdvancedResult)

# 프롬프트 작성
# partial(~~~~) : pydantic_parser의 필드를 참고해서 작성
prompt = ChatPromptTemplate.from_messages([
    ('system', '당신은 CX(고객경험) 분석가입니다.\n\n {format_instructions}'),
    ('human', '{text}'),
]).partial(format_instructions = pydatic_parser.get_format_instructions())

chain = prompt | watson_llm | pydatic_parser
result = chain.invoke({
    "text" : "배송은 빠른데 포장이 너무 허술해요."
})

print(result.sentiment) 
print(result.score) 
print(result.summary) 
print(f'카테고리 : {result.category}, 조치 : {result.action}')
```

#### RunnableLambda
- chain => RunnableSwquence 생성하기 때문에 일반 파이썬 함수 바로 연결 불가
- 일반 함수를 감싸는 함수

```python
def upper(text):
    return text.upper()

prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 언어 전문가입니다."),
    ("human", "{topic}에 대해 설명해줘"),
])

# chain = prompt | watson_llm | StrOutputParser() | upper()
chain = prompt | watson_llm | StrOutputParser() | RunnableLambda(upper)

print(chain.invoke({"topic":"python"}))
```

```python
# 후처리 함수
def remove_newline(text):
    return text.replace("\n"," ")

prompt = ChatPromptTemplate.from_template("{question}에 대해 답변해줘")

chain = prompt | watson_llm | StrOutputParser() | RunnableLambda(remove_newline)

print(chain.invoke({"question":"파이썬의 리스트"}))
```

#### RunnablePassThrough

```python
prompt = ChatPromptTemplate.from_template("{question}에 대해 답변해줘")

chain = (
    {
    "question": RunnablePassthrough(),                                      # question변수에 들어온 값이 들어옴
    "answer": prompt | watson_llm | StrOutputParser()                       # question에 대한 AI의 답변 값이 들어옴
    } | RunnableLambda(lambda x:f"Q{x['question']}\nA:{x['answer']}")       # 결과 값이 함수에 의해 실행됨
)

result =chain.invoke("파이썬이란?")

print(result)
```

#### RunnableParallel
- 여러 chain을 호출 할 때 사용
- 실제 병렬처리 성능은 컴퓨터 자원(ex. CPU, GPU, RAM 등)에 영향을 받음

```python
parser = StrOutputParser()

# 요약
summary_chain = ChatPromptTemplate.from_template("다음 텍스트를 한 문장으로 요약해줘:{text}") | watson_llm | parser

# 키워드 추출

keyword_chain = ChatPromptTemplate.from_template("다음 텍스트의 핵심 키워드 3개를 콤마로 나열해줘:{text}") | watson_llm | parser

parallel_chain = RunnableParallel(summary=summary_chain, keyword=keyword_chain)

result = parallel_chain.invoke({'text':'인공지능은 인간의 지능을 모방하는 기술로.....'})
print("요약", result['summary'])
print("키워드", result['keyword'])
```

#### .assign() : dict를 반환하는 결과에 붙혀서 사용 가능

```python
# {"text": LLM 답변}
analysis_chain = ChatPromptTemplate.from_template("다음 리뷰를 분석해줘:{review}") | watson_llm | parser | {"text":RunnablePassthrough()}

full_chain = (RunnablePassthrough.assign(summary = ChatPromptTemplate.from_template("{review} 한 줄 요약") | watson_llm | parser)
                    .assign(sentiment =ChatPromptTemplate.from_template("{review} 감정: 긍정/부정/중립 중 하나만") | watson_llm | parser)
                    .assign(analysis = analysis_chain))

result = full_chain.invoke({"review":"정말 맛있는 음식이었어요"})
print(result)
```

#### RunnableWithMessageHistory (자동 이력 관리)

- InMemoryChatMessageHistory()
    - 메모리에 대화 저장 후 대화 기록 관리
    

```python
history = InMemoryChatMessageHistory()

def chat(user_input):
    history.add_user_message(user_input)
    response = watson_llm.invoke(history.messages) # 전체 대화 이력 전송 => context =>한도 초과시 In-Memory OverFlow 발생가능
    history.add_ai_message(response.content)
    return response.content

print(chat("파이썬이란?"))
print(chat("방금 말한 내용의 장점 3가지는?"))
print(chat("그 중 첫 번째 장점에 대한 예시 코드 작성해줘"))
print(chat("두 번째 장점에 대한 예시 코드 작성해줘"))

for m in history.messages:
    print(f"[{m.type}] {m.content[:40]}....")
```

```python
# 여러 사람과 대화하는 부분 저장 => 세션별 저장

# 세션별 이력 저장소

store = {}

def get_session_history(session_id): # -> BaseChatMessageHistory()
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory() # 한 사람당 하나씩 부여
    return store[session_id]

# 프롬프트 생성
prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 {role}입니다. 한국어로 대답하세요."),
    MessagesPlaceholder(variable_name="history"), # 대화 이력 삽입 위치
    ("human", "{input}")
])

#LLM 생성
#체인 생성
chain = prompt | watson_llm | StrOutputParser()

# 1.session_id 이력 조회
# 2. history 키에 이력 삽입 -> prompt -> llm 실행 -> 대화 이력을 history에 추가

with_history = RunnableWithMessageHistory(
    chain, get_session_history, input_messages_key="input", history_messages_key="history")

# 세션 생성

cfg_a = {"configurable":{"session_id":"user_alice"}}
cfg_b = {"configurable":{"session_id":"user_bob"}}

# Alice 대화
res1 = with_history.invoke({"role":"파이썬 튜터", "input":"안녕"}, config=cfg_a)
res2 = with_history.invoke({"role":"파이썬 튜터", "input":"내 이름은 앨리스야"}, config=cfg_a)
res4 = with_history.invoke({"role":"파이썬 튜터", "input":"내 이름 기억해?"}, config=cfg_a)

# Bob 대화
res3 = with_history.invoke({"role":"파이썬 튜터", "input":"파이썬 변수 타입은?"}, config=cfg_b)

print("Alice",res2)
print("Alice",res4)
#print("Bob",res3)
```

- 대화가 길어지면 이력 전체를 LLM에 넘기는 경우 Context Window를 초과
- Memory에 어떤정보를 얼마나 보존할 것인지 결정
- 전략
    - 최근 K턴만 유지
    - 오래된 대화를 요약

```python
# 오래된 대화 요약

summary_prompt = ChatPromptTemplate.from_messages([
    ("system", "다음 대화 내용을 핵심만 3문장 이내의 한국어로 요약하세요"),
    ("human","{conversation}"),
])

summary_chain = summary_prompt | watson_llm | StrOutputParser()

# 요약기능이 포함된 class
class SummarizedChatHistory(InMemoryChatMessageHistory):
    """대화가 max_turns 초과시 오래된 메세지 요약하기"""
    max_turns:int = Field(default=6)
    summary:str = Field(default="")

    def _maybe_summarize(self):
        if len(self.messages) > self.max_turns * 2:
            # 오래된 내용 찾기
            cutoff = len(self.messages) // 2
            old_msgs = self.messages[:cutoff]
            new_msgs = self.messages[cutoff:]

            # 요약할 텍스트 구성
            conv_text = "\n".join(f"{'사용자' if isinstance(m, HumanMessage) else 'AI'}:{m.content}" for m in old_msgs)

            # 기존 요약 존재시 함께 포함
            if self.summary:
                conv_text = f"[이전 요약]\n{self.summary}\n\n[새 대화]\n{conv_text}"

            # 요약 실행
            self.summary = summary_chain.invoke({'conversation':conv_text})

            # 오래된 메세지 제거 후 요약본 SystemMessages 앞에 삽입
            self.messages.clear()
            self.messages.append(SystemMessage(content=f"[이전 대화 요약]\n{self.summary}"))
            self.messages.extend(new_msgs)

    def add_message(self, message):
        super().add_message(message)
        self._maybe_summarize()

store = {}
def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = SummarizedChatHistory(max_turns=6)

    return store[session_id]

# 프롬프트 생성
prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 친절한 AI 어시스턴트입니다. 한국어로 대답하세요."),
    MessagesPlaceholder(variable_name="history"), # 대화 이력 삽입 위치
    ("human", "{input}")
])

#LLM 생성
#체인 생성
chain = prompt | watson_llm | StrOutputParser()

# 1.session_id 이력 조회
# 2. history 키에 이력 삽입 -> prompt -> llm 실행 -> 대화 이력을 history에 추가

with_history = RunnableWithMessageHistory(
    chain, get_session_history, input_messages_key="input", history_messages_key="history")

# 세션 생성
cfg_a = {"configurable":{"session_id":"user_alice"}}

# 임의 대화 생성
questions = [
    "파이썬이란?",
    "방금 설명한 파이썬의 장점은?",
    "단점은?",
    "어떤 분야에서 많이 쓰이니?",
    "입문자에게 추천하는 학습 순서는?",
    "좋은 파이썬 입문서 알려줘",
    "무료로 배울 수 있는 사이트는?",
    "이전에 내가 뭘 물어봤는지 기억해?",
]

for i, q in enumerate(questions , 1):
    response = with_history.invoke({"input":q}, config=cfg_a)
    print(f"\n{i}턴 Q:{q}")
    print(f"\n      A:{response[:60]}....")

    # 요약 여부
    history = get_session_history("user_alice")
    if history.summary:
        print(f"\n 요약처리! 현재 메세지 수 {len(history.messages)}")
        print(f"\n 요약 내용 {history.summary[:80]}....")
```
